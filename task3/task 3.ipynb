```python
cap_a = 4
cap_b = 3
target = 2

checked_states = []

def water_jug_dfs(a, b):

    if (a, b) in checked_states:
        return False

    checked_states.append((a, b))

    if a == target or b == target:
        print("Target reached at state:", (a, b))
        return True

    if (cap_a, b) not in checked_states:
        print("Action: Fill Jug A")
        if water_jug_dfs(cap_a, b):
            return True

    if (a, cap_b) not in checked_states:
        print("Action: Fill Jug B")
        if water_jug_dfs(a, cap_b):
            return True

    if (0, b) not in checked_states:
        print("Action: Empty Jug A")
        if water_jug_dfs(0, b):
            return True

    if (a, 0) not in checked_states:
        print("Action: Empty Jug B")
        if water_jug_dfs(a, 0):
            return True

    move = min(a, cap_b - b)
    next_a = a - move
    next_b = b + move

    if (next_a, next_b) not in checked_states:
        print("Action: Transfer from Jug A to Jug B")
        if water_jug_dfs(next_a, next_b):
            return True

    move = min(b, cap_a - a)
    next_a = a + move
    next_b = b - move

    if (next_a, next_b) not in checked_states:
        print("Action: Transfer from Jug B to Jug A")
        if water_jug_dfs(next_a, next_b):
            return True

    return False

water_jug_dfs(0, 0)
```
